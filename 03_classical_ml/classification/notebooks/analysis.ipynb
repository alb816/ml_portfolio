## 3. EDA
На данном этапе исследуем рапределение целевой переменной и взаимосвязь основных признаков с выживаемостью пассажиров.

### 3.1 Распределение целевой переменной
Сначала посмотрим на распределение классов целевой переменной `survived`.

```python
X_train['Survived'].value_counts()
```

```python
X_train['Survived'].value_counts().plot(kind='bar')
plt.title('Survival Distribution')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

### Вывод
Целевая переменная имеет умеренный дисбаланс, поэтому при оценке моделей не следует ориентироваться только на Accuracy, а добавить также оценки Precision, Recall и F1-score.

### 3.2 Влияние пола на выживаемость
Проверим существует ли зависимость между полом пасажира и его выживаемостью.

```python
X_train.groupby(['Sex'])['Survived'].mean().plot(kind='bar')
plt.title('Survival Rate by Sex')
plt.ylabel('Rate')
plt.show()
```

```python
X_train.groupby(['Sex'])['Survived'].mean()
```

### Вывод 
Женщин в выборке выживало в среднем значительно больше, чем мужчин. 

Это значит что данный признак потенциально полезен для предсказания, поэтому он будет рассмотрен при построении модели.

### 3.3 Влияние класса пассажира на выживаемость

```python
pd.crosstab(
    X_train['Pclass'],
    X_train['Survived'],
    normalize='index'
)
```

```python
sns.countplot(
    data=X_train,
    x='Pclass',
    hue='Survived'
)

plt.title('Выживаемость в зависимости от класса пассажира')
plt.xlabel('Класс пассажира')
plt.ylabel('Кол-во пассажиров')
plt.show()
```
