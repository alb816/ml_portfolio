Для начала рассмотрим, как матрица линейного отображения(оператор) влияет на обычные векторы

```python
import numpy as np
import matplotlib.pyplot as plt


basis = np.array([[1, 0], [0, 1]]) # стандартный базис линейного пространства R^2
A = np.array([[1, 2], [2, -2]]) # оператор линейного отображения R^2 -> R^2

# Вычисляем трансформированные векторы
transformed = A @ basis.T
```

```python
plt.figure(figsize=(6,6))

plt.quiver([0, 0], [0, 0], basis[0,0], basis[1,0], color='gray', scale=5, label='Original')
plt.quiver([0, 0], [0, 0], basis[0,1], basis[1,1], color='gray', scale=5)

plt.quiver([0, 0], [0, 0], transformed[0,0], transformed[1,0], color='r', scale=5, label='Transformed i')
plt.quiver([0, 0], [0, 0], transformed[0,1], transformed[1,1], color='b', scale=5, label='Transformed j')

plt.xlim(-5, 5); plt.ylim(-5, 5); plt.grid(True); plt.legend()
plt.show()
```

Матрица повлияла не только на размеры векторов, но и на их направления.
