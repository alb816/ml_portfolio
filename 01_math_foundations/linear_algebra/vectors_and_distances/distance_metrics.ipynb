```python
import numpy as np


def l2_norm(x, y):
    "Евклидово расстояние"
    dist = np.sqrt(np.sum((x - y)**2))
    return dist

def l1_norm(x, y):
    "Расстояние Манхэттена"
    dist = np.sum(np.abs(x - y))
    return dist

def cosine_similarity(x, y):
    norm_x = np.linalg.norm(x)  # norm(x) = l2_norm(x) = sqrt(sum(x**2))
    norm_y = np.linalg.norm(y)
    if norm_x == 0 or norm_y == 0:
        return 0.0
    
    similarity = np.dot(x, y) / (norm_x * norm_y) # = cos w -- косинус угла между векторами x, y
    return similarity

def cosine_distance(x, y):
    return 1 - cosine_similarity(x, y)

def chebyshev_distance(x, y):
    dist = np.max(np.abs(x - y))
    return dist

def minkowski_distance(x, y, p=1):
    dist = np.sum(np.abs(x - y)**p)**(1/p)
    return dist
```

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
 
print(f"L2: {l2_norm(x, y)}")
print(f"L1: {l1_norm(x, y)}")
print(f"cosine: {cosine_distance(x, y)}")
print(f"chebyshev: {chebyshev_distance(x, y)}")
print(f"minkowski: {minkowski_distance(x, y, p=3)}")
```
