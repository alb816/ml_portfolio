# Базис, ранг и линейная независимость

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
v1 = np.array([1, 0])
v2 = np.array([0, 1])

vectors = np.column_stack([v1, v2])
rank = np.linalg.matrix_rank(vectors)
vectors, rank
```

Ранг равен двум, поэтому векторы v1, v2 линейно независимы. 
Они составляют базис в пространстве $ R^2 $.

```python
plt.figure()
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1,color='r') # координаты начала вектора: (X=2, Y=1). U=v1[0], V=v1[1]
                                                                                 # координаты конца вектора: (X+U, Y+V)
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='b')
plt.xlim(-1, 3) # ограничение длины оси X
plt.ylim(-1, 3)
plt.grid()
plt.title("Линейно независимые векторы")
plt.show()
```

```python
v1 = np.array([1, 1])
v2 = np.array([2, 2])

vectors = np.column_stack([v1, v2])
rank = np.linalg.matrix_rank(vectors)
rank
```

Так как ранг равен 1, то значит v1 и v2 линейно зависимы.

```python
plt.figure()
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1,color='r')
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='b')
plt.xlim(-1, 3) # ограничение длины оси X
plt.ylim(-1, 3)
plt.grid()
plt.title("Линейно зависимые векторы")
plt.show()
```

```python
A = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 1]
])

rank = np.linalg.matrix_rank(A)
rank
```

Ранг матрицы равен 2, значит кол-во базисных векторов равно 2. Т.е. размерность пространства, которое они порождают так же равно 2.
